```python
import kagglehub

path = kagglehub.dataset_download("abuzarkhaaan/vehicletest2")
print("Path to dataset files:", path)
```

```python
import os

for root, dirs, files in os.walk(path):
    level = root.replace(path, '').count(os.sep)
    indent = '  ' * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = '  ' * (level + 1)
    for f in files[:5]:  # cap at 5 files per folder so it doesn't flood output
        print(f"{subindent}{f}")
    if len(files) > 5:
        print(f"{subindent}... and {len(files)-5} more files")
```

```python
data_yaml_path = None
for root, dirs, files in os.walk(path):
    for f in files:
        if f in ("data.yaml", "data.yml"):
            data_yaml_path = os.path.join(root, f)
            break

print("Found data.yaml at:", data_yaml_path)
if data_yaml_path:
    with open(data_yaml_path) as f:
        print(f.read())
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
data_yaml_path="/root/.cache/kagglehub/datasets/abuzarkhaaan/vehicletest2/versions/1/data.yaml"
model.train(
    data=data_yaml_path,          # from Cell 5, or hardcode the path once you've confirmed it
    epochs=100,
    imgsz=640,
    batch=16,
    patience=20,
    project="/content/drive/MyDrive/Colab Notebooks/cv practice/vehicle_tracking_speed_estimation/pk_vehicles_det_runs",  # mount Drive first if you want persistence
    name="pk_vehicles",
    degrees=5,
    hsv_h=0.015,
    hsv_v=0.4,
    mosaic=1.0,
    close_mosaic=10,
    translate=0.1,
    scale=0.5,
)
```

```python
from ultralytics import YOLO
model = YOLO("/content/drive/MyDrive/Colab Notebooks/cv practice/vehicle_tracking_speed_estimation/pk_vehicles_det_runs/pk_vehicles-2/weights/best.pt")
metrics = model.val()
print(metrics.box.maps)  # per-class mAP50-95 array
```

```python
import os

base = "/root/.cache/kagglehub/datasets/abuzarkhaaan/vehicletest2/versions/1"
print(os.listdir(base))  # confirm folder structure first
```

```python
labels_dir = f"{base}/valid/labels"  # update if the printed structure above differs

segment_style = []
box_style = []

for label_file in os.listdir(labels_dir):
    has_segment = False
    with open(os.path.join(labels_dir, label_file)) as f:
        for line in f:
            if len(line.split()) > 5:
                has_segment = True
                break
    if has_segment:
        segment_style.append(label_file)
    else:
        box_style.append(label_file)

print(f"Segment-style label files: {len(segment_style)}")
print(f"Box-style label files: {len(box_style)}")

# Now check which classes are overrepresented in the segment-style files
from collections import Counter
class_counts_segment = Counter()
for label_file in segment_style:
    with open(os.path.join(labels_dir, label_file)) as f:
        for line in f:
            class_id = int(line.split()[0])
            class_counts_segment[class_id] += 1

print("Class distribution in segment-style labels:", class_counts_segment)
```
